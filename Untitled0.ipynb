```python
columnas = ['price' , 'miimun_nights', 'number of reviews' , 'reviews_per_month' , 'calculated_host_lidting_count' , 'availability_365']

data[columnas].describe()
```

```python
#Remplacemos los valores nulos, si existen
#Si hay valores nulos, podrías remplazarlos con la media (ejemplo):
data = data.fillna(data.mean())
```

```python
#Anuncios sin reseñas
print('Anuncios sin reseñas:', sin_reseñas.sum())
```
